```python
from faster_whisper import WhisperModel
```

```python
model = WhisperModel(
    "../model/",
    device="cpu",
    compute_type="int8"
)
```

```python
segments, _ = model.transcribe(
    "../audio/audio21.wav",
    language="pt",
    condition_on_previous_text=False,
)

full_text = " ".join([segment.text for segment in segments])
```
